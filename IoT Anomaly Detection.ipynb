```python
benign_df = pd.read_csv('BenignTraffic3.pcap.csv')
print(benign_df.shape)
```

```python
print(benign_df.columns.tolist())
```

```python
df['label'] = 'Attack'
```

```python
benign_df['label'] = 'Benign'
```

```python
combine_df = pd.concat([df,benign_df],ignore_index=True)
print(combine_df.shape)
```

```python
print(combine_df['label'].value_counts())
```

```python
X = combine_df.drop(columns=['label'])
y = combine_df['label']
```

```python
X = pd.get_dummies(X, drop_first=True)
```

```python
print(X.isnull().sum().sum())
```

```python
X = X.fillna(0)
```

```python
from sklearn.ensemble import IsolationForest
from sklearn.model_selection import train_test_split
```

```python
benign_X = X[y == 'Benign']
attack_X = X[y == 'Attack']
```

```python
train_X, test_benign_X = train_test_split(benign_X, test_size=0.3, random_state=42)
```

```python
model = IsolationForest(contamination=0.05, random_state=42)
model.fit(train_X)
```

```python
from sklearn.metrics import classification_report
```

```python
test_X = pd.concat([test_benign_X, attack_X])
test_y = ['Benign'] * len(test_benign_X) + ['Attack'] * len(attack_X)
```

```python
predictions = model.predict(test_X)
```

```python
df['label'] = 'Attack'
```

```python
pred_labels = ['Attack' if p == -1 else 'Benign' for p in predictions]
```

```python
print(classification_report(test_y, pred_labels))
```

```python
import pandas as pd

# Load the vulnerability scan file - replace with your exact filename/path
vuln_df = pd.read_csv('VulnerabilityScan.pcap.csv')
vuln_df['label'] = 'Attack'

# Combine with benign data
combined_df3 = pd.concat([benign_df, vuln_df], ignore_index=True)

print(combined_df3.shape)
print(combined_df3['label'].value_counts())
```

```python
X3 = combined_df3.drop(columns=['label'])
y3 = combined_df3['label']

X3 = pd.get_dummies(X3, drop_first=True)
X3 = X3.fillna(0)
```

```python
from sklearn.ensemble import IsolationForest
from sklearn.model_selection import train_test_split

benign_X3 = X3[y3 == 'Benign']
attack_X3 = X3[y3 == 'Attack']

train_X3, test_benign_X3 = train_test_split(benign_X3, test_size=0.3, random_state=42)

model3 = IsolationForest(contamination=0.05, random_state=42)
model3.fit(train_X3)
```

```python
from sklearn.metrics import classification_report

test_X3 = pd.concat([test_benign_X3, attack_X3])
test_y3 = ['Benign'] * len(test_benign_X3) + ['Attack'] * len(attack_X3)

predictions3 = model3.predict(test_X3)
pred_labels3 = ['Attack' if p == -1 else 'Benign' for p in predictions3]

print(classification_report(test_y3, pred_labels3))
```

```python
# Take a random sample of attack rows to match benign class size roughly
attack_X3_sample = attack_X3.sample(n=len(test_benign_X3), random_state=42)

test_X3_balanced = pd.concat([test_benign_X3, attack_X3_sample])
test_y3_balanced = ['Benign'] * len(test_benign_X3) + ['Attack'] * len(attack_X3_sample)

predictions3_balanced = model3.predict(test_X3_balanced)
pred_labels3_balanced = ['Attack' if p == -1 else 'Benign' for p in predictions3_balanced]

print(classification_report(test_y3_balanced, pred_labels3_balanced))
```

```python
import pandas as pd

# Load the MITM ArpSpoofing file - replace with your exact filename/path
mitm_df = pd.read_csv('MITM-ArpSpoofing.pcap.csv')
mitm_df['label'] = 'Attack'

combined_df4 = pd.concat([benign_df, mitm_df], ignore_index=True)

print(combined_df4.shape)
print(combined_df4['label'].value_counts())
```

```python
from sklearn.ensemble import IsolationForest
from sklearn.model_selection import train_test_split

benign_X4 = X4[y4 == 'Benign']
attack_X4 = X4[y4 == 'Attack']

train_X4, test_benign_X4 = train_test_split(benign_X4, test_size=0.3, random_state=42)

model4 = IsolationForest(contamination=0.05, random_state=42)
model4.fit(train_X4)
```

```python
from sklearn.metrics import classification_report

# Balance the attack sample size to match benign test set size
attack_X4_sample = attack_X4.sample(n=len(test_benign_X4), random_state=42)

test_X4 = pd.concat([test_benign_X4, attack_X4_sample])
test_y4 = ['Benign'] * len(test_benign_X4) + ['Attack'] * len(attack_X4_sample)

predictions4 = model4.predict(test_X4)
pred_labels4 = ['Attack' if p == -1 else 'Benign' for p in predictions4]

print(classification_report(test_y4, pred_labels4))
```
